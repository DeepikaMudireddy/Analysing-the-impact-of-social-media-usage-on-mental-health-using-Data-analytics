## **WP2: BEHAVIORAL ANALYSIS**

**1.Is there a relationship between the amount of time spent on social media and mental health outcomes?**

**Objective**: To examine whether spending more time on social media is associated with higher depression levels.

```python
df['hours_per_day'].value_counts()
```

```python
df.columns
```

```python
plt.figure(figsize=(10,6))

sns.boxplot(
    x='hours_per_day',
    y='18._how_often_do_you_feel_depressed_or_down?',
    data=df
)

plt.title("Relationship Between Social Media Usage and Depression")
plt.xlabel("Hours Spent on Social Media Per Day")
plt.ylabel("Depression Score")
plt.xticks(rotation=45)

plt.show()
```

**Report:**
The box plot shows that people with high levels of everyday use of social media had higher scores of depression than those who used it less. A higher median depression score for increased daily usage would indicate that greater social media usage is linked with increased depression symptoms.

**2.Are more frequent social comparisons associated with greater anxiety and depression?**

**Objective**: To understand whether comparing oneself with others on social media is related to depression.

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    x='15._on_a_scale_of_1-5,_how_often_do_you_compare_yourself_to_other_successful_people_through_the_use_of_social_media?',
    y='18._how_often_do_you_feel_depressed_or_down?',
    data=df
)

plt.title("Social Comparison vs Depression")
plt.xlabel("Social Comparison Score")
plt.ylabel("Depression Score")

plt.show()
```

**Report:**
The scatter graph above represents the correlation between social comparison and depression. The results indicate that a positive relationship exists between a tendency to compare with others and depression scores.

**3. What are the negative effects of validation-seeking behavior on mental health?**

**Objective**: To examine the relationship between validation-seeking behavior and anxiety.

```python
plt.figure(figsize=(8,6))

sns.regplot(
    x='17._how_often_do_you_look_to_seek_validation_from_features_of_social_media?',
    y='13._on_a_scale_of_1_to_5,_how_much_are_you_bothered_by_worries?',
    data=df
)

plt.title("Validation Seeking vs Anxiety")
plt.xlabel("Validation Seeking")
plt.ylabel("Anxiety (Worries)")

plt.show()
```

**Report:** The regression plot shows the relationship between anxiety and validation seeking behavior. People who seek validation more often may also endorse more anxiety if they have an upward trend.

**4. Are there links between social media and sleep difficulties and impaired functioning during the day?**

**Objective**: To investigate whether increased social media use is associated with more sleep-related issues.

```python
plt.figure(figsize=(10,6))

sns.boxplot(
    x='hours_per_day',
    y='20._on_a_scale_of_1_to_5,_how_often_do_you_face_issues_regarding_sleep?',
    data=df
)

plt.title("Social Media Usage vs Sleep Issues")
plt.xlabel("Hours Spent on Social Media")
plt.ylabel("Sleep Issues")
plt.xticks(rotation=45)

plt.show()
```

**Report**:
This box plot shows the comparison of sleep problems by level of social media use. Regarding sleep problems, the results suggest that the more time spent on social media is linked to worse sleep quality in heavy users, as sleep issue scores were higher among heavy users.

**5. What are the social media activities that seem to be most closely linked to poor mental health markers?**

**Objective**: To identify which social media behaviors are most strongly associated with anxiety, depression, and sleep problems.

```python
analysis = df[[
'12._on_a_scale_of_1_to_5,_how_easily_distracted_are_you?',
'13._on_a_scale_of_1_to_5,_how_much_are_you_bothered_by_worries?',
'15._on_a_scale_of_1-5,_how_often_do_you_compare_yourself_to_other_successful_people_through_the_use_of_social_media?',
'17._how_often_do_you_look_to_seek_validation_from_features_of_social_media?',
'18._how_often_do_you_feel_depressed_or_down?',
'19._on_a_scale_of_1_to_5,_how_frequently_does_your_interest_in_daily_activities_fluctuate?',
'20._on_a_scale_of_1_to_5,_how_often_do_you_face_issues_regarding_sleep?'
]]
```
